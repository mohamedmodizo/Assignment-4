1. Bar Plot: Total Amount Spent per Campaign

```python
# Cleaning of currency column and convert to float
df["Amount Spent Cleaned"] = df["Amount Spent in INR"].replace('[\$,]', '', regex=True).astype(float)
```

```python
# Then group by Campaign Name to get total spend
campaign_summary = df.groupby("Campaign Name")["Amount Spent Cleaned"].sum().reset_index()
```

```python
# Ploting bar plot
"""
This is a bar graph
Ploting X againts Y
Analyses xyz
"""
plt.figure(figsize=(10, 6))
sns.barplot(data=campaign_summary, x="Amount Spent Cleaned", y="Campaign Name", palette="Blues_d")
plt.title("Total Amount Spent per Campaign")
plt.xlabel("Amount Spent (INR)")
plt.ylabel("Campaign Name")
plt.tight_layout()
plt.show()
```

2. Bar Plot: Click-Through Rate (CTR %) by Campaign

```python
# Group by campaign and then calculate average CTR
campaign_ctr = df.groupby("Campaign Name")["Click-Through Rate (CTR in %)"].mean().reset_index()
```

```python
# Plot CTR by Campaign graph
plt.figure(figsize=(10, 6))
sns.barplot(
    data=campaign_ctr,
    x="Click-Through Rate (CTR in %)",
    y="Campaign Name",
    palette="flare"
)
plt.title("Click-Through Rate (CTR %) by Campaign")
plt.xlabel("CTR (%)")
plt.ylabel("Campaign Name")
plt.tight_layout()
plt.show()
```

3. Bar Plot: Cost Per Result (CPR) by Campaign

```python
# Clean the 'Cost per Result (CPR)' column
df['CPR Cleaned'] = df['Cost per Result (CPR)'].replace('[\$,]', '', regex=True).astype(float)
```

```python
# Group by campaign and then calculate average CPR
campaign_cpr = df.groupby("Campaign Name")["CPR Cleaned"].mean().reset_index()
```

```python
# Plot CPR by Campaign graph
plt.figure(figsize=(10, 6))
sns.barplot(
    data=campaign_cpr,
    x="CPR Cleaned",
    y="Campaign Name",
    palette="mako"
)
plt.title("Cost Per Result (CPR) by Campaign")
plt.xlabel("Cost Per Result (INR)")
plt.ylabel("Campaign Name")
plt.tight_layout()
plt.show()
```

4. Scatter Plot: Amount Spent vs Clicks

```python
# Clean 'Amount Spent in INR' column
df['Amount Spent Cleaned'] = df['Amount Spent in INR'].replace('[\$,]', '', regex=True).astype(float)
```

```python
# Group by Campaign to get total amount spent and clicks
campaign_spend_clicks = df.groupby("Campaign Name").agg({
    "Amount Spent Cleaned": "sum",
    "Clicks": "sum"
}).reset_index()
```

```python
# Plot Scatter Plot
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=campaign_spend_clicks,
    x="Amount Spent Cleaned",
    y="Clicks",
    hue="Campaign Name",
    s=100,
    palette="tab10"
)
plt.title("Amount Spent vs Clicks by Campaign")
plt.xlabel("Amount Spent (INR)")
plt.ylabel("Total Clicks")
plt.tight_layout()
plt.show()
```

5. Bubble Plot: CTR vs CPR

```python
# Clean the 'Amount Spent' and 'Cost per Result' columns
df['Amount Spent Cleaned'] = df['Amount Spent in INR'].replace('[\$,]', '', regex=True).astype(float)
df['CPR Cleaned'] = df['Cost per Result (CPR)'].replace('[\$,]', '', regex=True).astype(float)
```

```python
# Group by campaign and compute required metrics
campaign_bubble = df.groupby("Campaign Name").agg({
    "Click-Through Rate (CTR in %)": "mean",
    "CPR Cleaned": "mean",
    "Amount Spent Cleaned": "sum"
}).reset_index()
```

```python
# Create the bubble plot
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=campaign_bubble,
    x="Click-Through Rate (CTR in %)",
    y="CPR Cleaned",
    size="Amount Spent Cleaned",
    hue="Campaign Name",
    sizes=(100, 1000),
    alpha=0.7,
    legend=True
)
plt.title("CTR vs CPR (Bubble Size = Amount Spent)")
plt.xlabel("Click-Through Rate (%)")
plt.ylabel("Cost Per Result (INR)")
plt.tight_layout()
plt.show()
```
